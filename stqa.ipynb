# Practical 7: Load Testing a Web Application Using JMeter

## 1. Aim

To conduct load testing on a web application using Apache JMeter to measure its performance, response time, and stability when multiple users access it simultaneously.

## 2. Software required

- Apache JMeter
- Java (a compatible JDK)
- A web browser
- A web application or test website that you own or have permission to test

## 3. Theory

Load testing is a type of performance testing used to check how a web application behaves when many users access it at the same time.

It helps identify slow response times, errors, and performance bottlenecks.

Important terms:

- Threads (Users): Number of virtual users accessing the application.
- Ramp-up period: Time JMeter takes to start all virtual users.
- Loop count: Number of times each user repeats the test.
- Response time: Time taken by the server to respond to a request.
- Throughput: Number of requests processed per unit of time.
- Error rate: Percentage of requests that fail.

## 4. Procedure

### Step 1: Download and install JMeter

1. Install a Java version compatible with your JMeter release.
2. Download Apache JMeter from the [official JMeter website](https://jmeter.apache.org/download_jmeter.cgi?utm_source=chatgpt.com).
3. Extract the downloaded ZIP or TAR file.
4. Open the extracted folder and go to `bin`.
5. Start JMeter:
   - Windows: Run `jmeter.bat`
   - Linux/macOS: Run `./jmeter`

See the official [JMeter installation guide](https://jmeter.apache.org/usermanual/get-started.html?utm_source=chatgpt.com).&#x20;

## 6. Result analysis

- Low response time: Requests are being processed relatively quickly.
- High response time: The application may be experiencing delays under load.
- High error percentage: Some requests are failing and should be investigated.
- Low throughput: The server may not be processing requests efficiently, or the test setup may be limiting the load.

Compare the measurements against the application's performance requirements. There is no single response time that is acceptable for every application.

# Practical: Testing SOAP Web Services for Functionality and Fault Checking

## 1. Aim

To test a SOAP (Simple Object Access Protocol) web service using SoapUI to verify its functionality, validate responses, and check how the service handles faults and invalid requests.

## 2. Software required

- SoapUI (Open Source)
- Internet connection, if using a public sample service
- A SOAP WSDL file (Web Services Description Language) or a SOAP service endpoint

## 3. Theory

SOAP is a protocol used for exchanging structured information between applications, commonly using XML messages.

SOAP web service testing checks whether:

- The service returns the expected response for valid requests.
- The response contains the correct data.
- Invalid inputs are handled properly.
- SOAP Fault messages are returned when errors occur.
- The service behaves correctly when required fields are missing or values are invalid.

A SOAP Fault is a structured error response that provides information about a problem processing a SOAP request.

## 4. Procedure

### Step 1: Install and open SoapUI

Download SoapUI from the official website: [SoapUI Downloads](https://www.soapui.org/downloads/soapui/?utm_source=chatgpt.com).

Assertions automatically check whether the returned response matches expectations.

For the Contains assertion, use `30` as the expected text. For a more precise check, use an XPath assertion that verifies the value of `AddResult`.

SoapUI supports response, schema, and SOAP Fault assertions for functional and negative testing.&#x20;

## 5. Perform fault checking

Fault checking verifies how the service behaves when it receives invalid or malformed requests.

Try these test cases in a test environment.

| Test case         | Input or action                  | Expected result                                          |
| ----------------- | -------------------------------- | -------------------------------------------------------- |
| Valid request     | `10 + 20`                        | Returns `30`                                             |
| Zero values       | `0 + 0`                          | Returns `0`                                              |
| Negative values   | `-5 + 3`                         | Returns `-2`                                             |
| Missing parameter | Remove `intA`                    | Validation error or SOAP Fault, depending on the service |
| Invalid data type | Send `abc` instead of an integer | SOAP Fault or input validation error                     |
| Malformed XML     | Remove a closing XML tag         | Request rejected or parsing error                        |

Important: A fault is not guaranteed for every invalid input. The actual result depends on the service implementation, SOAP version, and validation rules. Record the response you actually receive.

### How to check a SOAP Fault

A SOAP 1.1 fault may look like this:

```
<soap:Envelope
    xmlns:soap="http://schemas.xmlsoap.org/soap/envelope/">
  <soap:Body>
    <soap:Fault>
      <faultcode>soap:Client</faultcode>
      <faultstring>Invalid request</faultstring>
    </soap:Fault>
  </soap:Body>
</soap:Envelope>
```

This is an illustrative example, not a guaranteed response from the Calculator service.

To verify a fault in SoapUI:

1. Send an invalid request.
2. Inspect the response body and HTTP status.
3. Look for the SOAP `Fault` element and its fault code or reason.
4. In the negative test case, add a SOAP Fault assertion if a fault is expected.
5. Record whether the actual response matches the expected behavior.

## 3. Theory

Continuous testing means automatically running software tests as part of the development process. When a developer updates the source code, the pipeline can:

1. Retrieve the latest code from GitHub.
2. Install required dependencies.
3. Run automated tests.
4. Report whether the tests passed or failed.

A Jenkins pipeline is commonly defined in a file named `Jenkinsfile`, stored alongside the application code in Git.&#x20;

# Practical: Web Application Vulnerability Assessment and Basic Penetration Testing

## 1. Aim

To identify and document common security vulnerabilities in a web application and simulate a basic penetration-testing scenario in an authorized test environment.

## 2. Software required

- OWASP ZAP — open-source web application security scanner.
- A web browser.
- A deliberately vulnerable practice application, such as OWASP Juice Shop.
- A report document or spreadsheet to record findings.

Official resources:

- OWASP ZAP — [Official website](https://www.zaproxy.org/?utm_source=chatgpt.com)
- OWASP Juice Shop — [Official project](https://owasp.org/www-project-juice-shop/?utm_source=chatgpt.com)

Safety note: Perform scans and penetration tests only on applications you own or have explicit permission to test. For this practical, use a local training application rather than a public website.

## 3. Theory

A vulnerability assessment identifies and evaluates security weaknesses in an application. Penetration testing goes further by safely testing whether identified weaknesses can be exploited.

Common web application vulnerabilities include:

- SQL Injection: Unsafe handling of user input in database queries.
- Cross-Site Scripting (XSS): Untrusted input being interpreted as executable browser code.
- Broken access control: Users being able to access resources they are not authorized to view.
- Security misconfiguration: Incorrect security settings or unnecessarily exposed information.
- Weak authentication: Inadequate login, password, or session protections.

A vulnerability should be documented with evidence, its potential impact, severity, and recommended remediation.

## 4. Procedure

### Step 1: Set up a vulnerable test application

Use OWASP Juice Shop as the practice target.

If Docker is installed, open a terminal and run:

```
docker run --rm -p 3000:3000 bkimminich/juice-shop
```

Open your browser and visit:

`http://localhost:3000`

You should see the Juice Shop application. Keep the terminal running while you test.

This setup exposes the practice application on your own computer. Do not expose it to the internet.

### Step 2: Install and open OWASP ZAP

1. Download ZAP from its [official download page](https://www.zaproxy.org/download/?utm_source=chatgpt.com).
2. Install and launch it.

### Step 4: Perform a basic security scan

For an authorized local target, you can use ZAP's automated scan or baseline scan. The baseline scan performs spidering and passive analysis without running the active attack scan.

For example, if you have the ZAP Docker image available, run:

```
docker run --rm \
  -t ghcr.io/zaproxy/zaproxy:stable \
  zap-baseline.py \
  -t http://host.docker.internal:3000 \
  -r zap-report.html
```

This hostname is commonly supported on Docker Desktop. On Linux, container-to-host networking may need different configuration. Make sure the target is reachable from the container before running the scan.

The generated report, `zap-report.html`, summarizes potential findings. A baseline scan is not a complete penetration test, and findings must be reviewed manually.

### Step 5: Simulate basic security checks

Perform these safe checks in your local test environment.

| Test                | What to check                                                        | Security concern                     |
| ------------------- | -------------------------------------------------------------------- | ------------------------------------ |
| Login validation    | Invalid credentials are rejected                                     | Authentication weakness              |
| Input handling      | Unexpected input is safely handled                                   | Injection or input-validation issues |
| Access control      | Restricted pages require authorization                               | Unauthorized access                  |
| HTTPS configuration | Sensitive traffic uses HTTPS in deployment                           | Data exposure in transit             |
| Security headers    | Review response headers                                              | Browser-side protection gaps         |
| Session cookies     | Check `Secure`, `HttpOnly`, and `SameSite` settings where applicable | Session theft or misuse risks        |

For example, in ZAP's response or alert details, inspect whether session cookies have appropriate flags. Do not assume that a missing header automatically means the application is exploitable; evaluate the application's context and deployment.

### Step 6: Review and validate findings

For each ZAP alert:

1. Record the alert name and affected URL.
2. Read the evidence and description.
3. Confirm whether the issue applies to the application.
4. Estimate the impact and severity.
5. Record a recommendation to fix it.

Do not treat every scanner alert as a confirmed vulnerability. Automated scanners can produce false positives and may miss vulnerabilities.

### Example finding

### VULN-01 — Missing security header

Example: Low

Needs validation

Observation: A scan reports that a response does not include a recommended security header.

Potential impact: Depending on the application and the missing header, browser protections may be reduced.

Recommendation: Review the applicable header, such as Content-Security-Policy, and configure a suitable policy after testing compatibility.

Evidence: Record the actual response header and the relevant ZAP alert.

This is an illustrative example, not a claim that your scan found this issue.

## 6. Remediation and retesting

After identifying an issue:

1. Fix the relevant application code or configuration.
2. Restart or redeploy the local test application if needed.
3. Repeat the same security check.
4. Record whether the issue is resolved.
5. Update the report with the retest date and result.
